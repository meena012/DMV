Problem Statement: Analyzing Sales Data from Multiple File Formats 
Dataset: Sales data in multiple file formats (e.g., CSV, Excel, JSON) 
Description: The goal is to load and analyze sales data from different file formats, including 
CSV, Excel, and JSON, and perform data cleaning, transformation, and analysis on the 
dataset. 

```python
import pandas as pd
import numpy as np
import json
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
csv_file = "sales_data_sample - sales_data_sample (2).csv"
excel_file = "Sample-Sales-Data (2).xlsx"
json_file = "pdf-data.json"

df_csv = pd.read_csv(csv_file)
df_excel = pd.read_excel(excel_file)

with open(json_file, "r", encoding="utf-8") as f:
    json_data = json.load(f)

print("CSV Data:")
display(df_csv.head())

```

```python
print("Excel Data:")
display(df_excel.head())

```

```python
print("Uploaded JSON type:", type(json_data))
print("Uploaded JSON keys:", list(json_data.keys()))
```

```python
print("CSV shape:", df_csv.shape)
print("\nCSV columns:")
print(df_csv.columns.tolist())

print("\nMissing values in CSV:")
print(df_csv.isnull().sum())

print("\nDuplicate rows in CSV:", df_csv.duplicated().sum())

print("\nExcel shape:", df_excel.shape)
print("\nMissing values in Excel:")
print(df_excel.isnull().sum())

print("\nCSV statistical summary:")
display(df_csv.describe(include="all"))
```

```python
# Remove duplicate rows
df_csv = df_csv.drop_duplicates().copy()
df_excel = df_excel.drop_duplicates().copy()

# Clean column names
df_csv.columns = df_csv.columns.str.strip()
df_excel.columns = df_excel.columns.str.strip()

# Convert sales columns to numeric
df_csv["SALES"] = pd.to_numeric(df_csv["SALES"], errors="coerce")
df_csv["QUANTITYORDERED"] = pd.to_numeric(
    df_csv["QUANTITYORDERED"], errors="coerce"
)

# Convert order dates
df_csv["ORDERDATE"] = pd.to_datetime(
    df_csv["ORDERDATE"], errors="coerce"
)

# Remove rows missing essential sales information
df_csv = df_csv.dropna(
    subset=["ORDERNUMBER", "SALES", "ORDERDATE", "PRODUCTLINE"]
)

# Remove invalid sales and quantities
df_csv = df_csv[
    (df_csv["SALES"] >= 0) &
    (df_csv["QUANTITYORDERED"] > 0)
].copy()

# Clean Excel data
df_excel["Year"] = pd.to_numeric(df_excel["Year"], errors="coerce")
df_excel["Value"] = pd.to_numeric(df_excel["Value"], errors="coerce")

df_excel = df_excel.dropna(
    subset=["Year", "Value", "Sales_Rep_Name"]
)

print("Data cleaning completed!")
print("Clean CSV shape:", df_csv.shape)
print("Clean Excel shape:", df_excel.shape)
```

```python
# Save sales data in JSON format
df_csv.to_json(
    "sales_data.json",
    orient="records",
    date_format="iso",
    indent=4
)

# Load the newly created JSON sales file
df_json = pd.read_json("sales_data.json")

print("Sales JSON loaded successfully!")
display(df_json.head())

print("JSON shape:", df_json.shape)
```

```python
# Annual sales from the CSV order dataset
csv_summary = (
    df_csv.groupby(df_csv["ORDERDATE"].dt.year)["SALES"]
    .sum()
    .reset_index()
)

csv_summary.columns = ["Year", "Sales"]
csv_summary["Source"] = "CSV Orders"

# Annual values from the Excel sales-representative dataset
excel_summary = (
    df_excel.groupby("Year")["Value"]
    .sum()
    .reset_index()
)

excel_summary.columns = ["Year", "Sales"]
excel_summary["Source"] = "Excel Sales Representatives"

# Combine the summaries
unified_data = pd.concat(
    [csv_summary, excel_summary],
    ignore_index=True
)

print("Unified annual summary:")
display(unified_data)
```

```python
#Descriptive statistics

print("Sales statistics:")
display(df_csv[["QUANTITYORDERED", "PRICEEACH", "SALES"]].describe())
```

```python
#Total sales and average order value
total_sales = df_csv["SALES"].sum()

order_totals = df_csv.groupby("ORDERNUMBER")["SALES"].sum()
average_order_value = order_totals.mean()

print("Total Sales: $", round(total_sales, 2))
print("Number of Orders:", df_csv["ORDERNUMBER"].nunique())
print("Average Order Value: $", round(average_order_value, 2))
```

```python
#Sales by product category
category_sales = (
    df_csv.groupby("PRODUCTLINE")["SALES"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)
```

```python
#Sales by year
yearly_sales = (
    df_csv.groupby(df_csv["ORDERDATE"].dt.year)["SALES"]
    .sum()
)

print(yearly_sales)
```

```python
#Sales by country
country_sales = (
    df_csv.groupby("COUNTRY")["SALES"]
    .sum()
    .sort_values(ascending=False)
)

display(country_sales.head(10))
```

```python
category_sales.plot(kind="bar", figsize=(9, 5))

plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Sales ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
yearly_sales.plot(kind="line", marker="o", figsize=(8, 5))

plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Sales ($)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
category_sales.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(8, 8)
)

plt.title("Sales Distribution by Product Line")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df_csv["SALES"])

plt.title("Distribution of Individual Sales Records")
plt.xlabel("Sales ($)")
plt.show()
```

```python
df_csv.to_csv("cleaned_sales.csv", index=False)
df_csv.to_excel("cleaned_sales.xlsx", index=False)
df_csv.to_json("cleaned_sales.json", orient="records", indent=4)

print("Cleaned sales data saved in CSV, Excel, and JSON formats!")
```
